# Compare matrix results
Filter within a dataset, compare one method across datasets, and inspect run coverage.

```python
from pathlib import Path

import pandas as pd

RUN_NAME = "all_qa_datasets_smoke"
DATASET = "qasper"
METHOD = "fixed_256"
run_dir = Path("outputs") / RUN_NAME
results = pd.read_csv(run_dir / "benchmark_metrics.csv")

within_dataset = results[results["Dataset"] == DATASET]
across_datasets = results[results["Method"] == METHOD]
coverage_matrix = results.pivot(
    index="Dataset", columns="Method", values="Status"
)
completed = results[results["Status"] == "completed"]
complete_method_counts = completed.groupby("Method")["Dataset"].nunique()
complete_methods = complete_method_counts[complete_method_counts == 4].index
macro_average = completed[completed["Method"].isin(complete_methods)].groupby(
    "Method"
)["EvidenceCoverage@10"].mean()

within_dataset, across_datasets, coverage_matrix, macro_average
```
