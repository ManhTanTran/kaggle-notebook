# Run the multi-dataset chunking benchmark
The notebook only selects configuration, calls the matrix runner, and reads artifacts.

```python
from pathlib import Path

import pandas as pd

from chunkbench.config.loader import load_config
from chunkbench.pipeline.matrix import ExperimentMatrixRunner

EXPERIMENT_CONFIG = "configs/experiments/all_qa_datasets_smoke.yaml"
config = load_config(EXPERIMENT_CONFIG)
runner = ExperimentMatrixRunner(config)
matrix = runner.expand()

print("datasets:", [item["name"] for item in config.datasets])
print("methods:", [item["name"] for item in config.methods])
print("matrix size:", len(matrix))

result = runner.run()
run_dir = Path(result["run_dir"])
metrics = pd.read_csv(run_dir / "benchmark_metrics.csv")
print("completed:", len(result["completed_runs"]))
print("failed:", len(result["failed_runs"]))
metrics.head()
```
