# Error analysis
Select a dataset and method, then inspect its persisted retrieval and evidence artifacts.

```python
import json
from pathlib import Path

import pandas as pd

RUN_NAME = "all_qa_datasets_smoke"
DATASET = "qasper"
METHOD = "fixed_256"
method_dir = Path("outputs") / RUN_NAME / DATASET / METHOD

retrieval = pd.read_csv(method_dir / "retrieval.csv")
coverage = [
    json.loads(line)
    for line in (method_dir / "evidence_coverage.jsonl").read_text(
        encoding="utf-8"
    ).splitlines()
]
metrics = json.loads((method_dir / "metrics.json").read_text(encoding="utf-8"))
retrieval.head(), coverage[:3], metrics
```
