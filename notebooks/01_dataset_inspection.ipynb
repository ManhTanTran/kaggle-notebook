# Dataset inspection
Select a registered dataset, load its canonical bundle, and inspect validation diagnostics.

```python
from pathlib import Path

from chunkbench.config.loader import load_yaml
from chunkbench.data.validation import validate_dataset
from chunkbench.registry.datasets import build_dataset_adapter

DATASET_NAME = "qasper"
config_path = Path(f"configs/datasets/fixtures/{DATASET_NAME}.yaml")
dataset_config = load_yaml(config_path)
adapter = build_dataset_adapter(DATASET_NAME, dataset_config)
bundle = adapter.load()
diagnostics = validate_dataset(bundle, dataset_config["validation"])

print(bundle.metadata)
print(diagnostics)
print("document example:", bundle.documents[0])
print("query example:", bundle.queries[0])
print("evidence example:", bundle.evidence[0])
```
